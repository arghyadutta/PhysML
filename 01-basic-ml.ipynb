Artificial Intelligence (AI): A simulation of human intelligence in machines that are programmed to think, reason, and learn. 

Types:

2. General AI: *Hypothetical* systems capable of performing any intellectual task a human can do.

Learning algorithms are computational methods used by machines to improve their performance on tasks over time based on experience (data).

-Reinforcement Learning: Learning through interaction with the environment.Examples: Game-playing AI (Q-Learning)


## Some examples of supervised ML

- Goal: make predictions from data
- Example: email spam filtering. An email (data instance) is classified as spam or not.
- Traditionally: one formulate rules to classify, but this has drawbacks:
  - Spam changes over time, so rules must be constantly rewritten.
  - Spammers can reverse-engineer fixed rules and design messages to evade them.
  - Hard-coded rules don't generalize easily to other languages/domains.
- Machine learning instead learns a program from past data rather than programming it by hand.
- This works when we have data instances with *known correct predictions* (labels), e.g. user-annotated spam/not-spam emails.
- A learning algorithm uses such labeled data to learn a *classifier* that predicts the correct *label* for new data.
- Other applications: web-search ranking, online ad placement, visual object recognition, face detection.

## Setup for supervised ML

- Training data comes in pairs of inputs $(\mathbf{x},y)$, where $\mathbf{x} \in \mathcal{R}^{d}$ is the input instance and $y \in \mathcal{C}$ is its label.
- The entire training data set is denoted:
$$
D = \left\{ (\mathbf{x}_{1},y_{1}),\ldots,(\mathbf{x}_{n},y_{n}) \right\} \subseteq \mathcal{R}^{d} \times \mathcal{C}
$$
- Notation:
  - $\mathcal{R}^{d}$: the $d$-dimensional feature space
  - $\mathbf{x}_{i}$: the input vector of the $i^{th}$ sample
  - $y_{i}$: the label of the $i^{th}$ sample
  - $\mathcal{C}$: the label space
- Data points $(\mathbf{x}_{i},y_{i})$ are randomly drawn from a unknown distribution $\mathcal{P}(X,Y)$.
- Goal: learn a function $h$ such that for a new pair $(\mathbf{x},y) \sim \mathcal{P}$, $h(\mathbf{x}) = y$ with high probability (or $h(\mathbf{x}) \approx y$).

### Examples of Label Spaces

- Binary classification: $\mathcal{C} = \{0,1\}$. For example, in spam filtering: emails are either spam (0) or not (1).
- Multi-class classification: $\mathcal{C} = \{1,2,\cdots,K\}$, $(K \geq 2)$. For example, in face classification, a person has a particular identity out of $K$ possibilities.
- Regression: $\mathcal{C} = \mathbb{R}$. For example, predicting tomorrow's temperature.

## Examples of feature vectors

- $\mathbf{x}_{i}$ is called a feature vector; each of its $d$ dimensions is a *feature* describing sample $i$.
- Patient data in a hospital: $\mathbf{x}_{i} = (x_{i}^{1},x_{i}^{2},\cdots,x_{i}^{d})$, e.g. $x_i^1$ = gender (0/1), $x_i^2$ = height (cm), $x_i^3$ = age (years), etc.
- Text document in bag-of-words format:
  - $\mathbf{x}_{i} = (x_{i}^{1},x_{i}^{2},\cdots,x_{i}^{d})$, where $x_i^{\alpha}$ = term frequency of word $\alpha$ in document $i$.
  - Example (from [Wikipedia](https://en.wikipedia.org/wiki/Bag-of-words_model)): 
    - `John likes to watch movies. Mary likes movies too. Mary also likes to watch football games.`
    - `BoW = {"John":1,"likes":3,"to":2,"watch":2,"movies":2,"Mary":2,"too":1,"also":1,"football":1,"games":1}`
  - Typically the feature vector is sparse.
- Images: features are pixel values.
  - $\mathbf{x}_{i} = (x_{i}^{1},x_{i}^{2},\cdots,x_{i}^{3k})$, where every group of 3 coordinates gives the RGB values of one pixel.
  - Typically the feature vector is dense.
  - A $7MP$ camera yields $7M \times 3 = 21M$ features.

## Hypothesis classes

- Before finding $h$, we must specify the type of function we're looking for (regression, decision tree, etc.). The set of possible functions is the hypothesis class.
- Specifying the hypothesis class is guided by *our understanding and assumptions* about the problem.
- There is no single ML algorithm that works best for every setting.

## Loss Functions

- Learning a hypothesis $h(\cdot)$ typically has two steps:
  1. Choose the hypothesis class $\mathcal{H}$ (the type of ML algorithm / set of learnable functions).
  2. Find the best $h \in \mathcal{H}$ (the actual learning/optimization process).
- We look for the $h$ that is 'simple' and makes the fewest mistakes on the training data.
- A loss function (or risk function) evaluates a hypothesis $h \in \mathcal{H}$ on training data and tells us how bad the hypothesis is.
  - Higher loss = worse; zero loss = perfect predictions.

### Zero-one loss

- Counts how many mistakes $h$ makes on the training set: loss of 1 per misprediction, 0 otherwise.
- training error = Normalized zero-one loss
$$
\mathcal{L}_{0/1}(h) = \frac{1}{n}\sum\limits_{i = 1}^{n}\delta_{h(\mathbf{x}_{i}) \neq y_{i}},\text{ where } \delta_{h(\mathbf{x}_{i}) \neq y_{i}} = \left\{ \begin{matrix}
{1,} & {\text{ if } h(\mathbf{x}_{i}) \neq y_{i}} \\
{0,} & \text{ o.w.}
\end{matrix} \right.
$$
- Common for evaluating classifiers, but rarely used to guide optimization since it's non-differentiable and non-continuous.


### Squared loss

- Typically used in regression:
$$
\mathcal{L}_{sq}(h) = \frac{1}{n}\sum\limits_{i = 1}^{n}(h(\mathbf{x}_{i}) - y_{i})^{2}
$$
- It's nonnegative and
- grows quadratically with the mispredictions, discouraging large errors.


### Absolute loss

- Also used in Regression
$$
\mathcal{L}_{abs}(h) = \frac{1}{n}\sum\limits_{i = 1}^{n}|h(\mathbf{x}_{i}) - y_{i}|
$$
- Grows *linearly* with mispredictions, making it more robust/suitable for noisy data (large mispredictions don't dominate the loss).

## Generalization

- Given a loss function, find $h$ minimizing the loss:
$$
h = \text{argmin}_{h \in \mathcal{H}}\mathcal{L}(h)
$$
- A major focus of ML is *how* to do this minimization efficiently.
- However, making sure that $h(\cdot)$ has low loss on training data $D$ does not ensure that it will work on unseen examples that are not in $D$? For example, consider the "memorizer" hypothesis:
$$
h(x) = \left\{ \begin{matrix}
{y_{i},} & {\text{ if } \exists(\mathbf{x}_{i},y_{i}) \in D \text{ s.t. } \mathbf{x} = \mathbf{x}_{i}} \\
{0,} & \text{ o.w.}
\end{matrix} \right.
$$
  - Achieves $0\%$ error on training data $D$, but performs horribly on samples not in $D$. This is overfitting.

## Train / Test splits

- To address overfitting, split $D$ into three subsets:
  - training data $D_{\text{TR}}$
  - validation data $D_{\text{VA}}$
  - test data $D_{\text{TE}}$
- For example, by $80\%$ / $10\%$ / $10\%$ (use most data for training, we need to build a good model and we can't afford losing data!)
- Choose $h(\cdot)$ based on $D_{\text{TR}}$, then evaluate on $D_{\text{TE}}$.
- Validation data is needed to check if $h$ overfits. That is, if the loss is large in validation set, we retrain, keeping the test data untouched.

**Remember**

- Spam filter: train on past data to predict future spam --> temporal splitting
- If not, splitting uniformly at random is often best.
- **Never split alphabetically or by feature values**.
- The test error (testing loss) approximates the true generalization error/loss.

## Summary

- Learning: $h^{\ast}( \cdot ) = \text{argmin}_{h( \cdot ) \in \mathcal{H}}\frac{1}{|D_{\text{TR}}|}\sum\limits_{(\mathbf{x},y) \in D_{\text{TR}}}\ell(\mathbf{x},y|h( \cdot ))$
- Evaluate learning:
$\epsilon_{\text{TE}} = \frac{1}{|D_{\text{TE}}|}\sum\limits_{(\mathbf{x},y) \in D_{\text{TE}}}\ell(\mathbf{x},y|h^{\ast}( \cdot ))$
- Generalization: If samples are drawn i.i.d. from the same distribution, the testing loss tells how good the model is. (It's hard to predict tomorrow's temperature in Greenland if all of your training data is from Guntur. Check distribution shift online. 
  - See this, for example: J. Wu, Distribution shifts in trustworthy machine learning, [AI Magazine 47, e70057 (2026)](https://onlinelibrary.wiley.com/doi/10.1002/aaai.70057).
- No free lunch: every ML algorithm makes assumptions about which hypothesis class $\mathcal{H}$ to choose. There is no single perfect $\mathcal{H}$ for all problems.
- A beautiful example from Prof. Weinberger: given $(\mathbf{x}_1,y_1)=(1,1)$, $(\mathbf{x}_2,y_2)=(2,2)$, $(\mathbf{x}_3,y_3)=(3,3)$, $(\mathbf{x}_4,y_4)=(4,4)$, $(\mathbf{x}_5,y_5)=(5,5)$:
  - What is $y$ when $\mathbf{x} = 2.5$? **Impossible to know without assumptions**.
